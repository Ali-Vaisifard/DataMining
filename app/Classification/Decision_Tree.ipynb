```python
# Import libraries
import pandas as pd
import numpy as np
import seaborn as sns 
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
# Generrating Random Data
np.random.seed(123)

X = np.random.randn(2000, 5)
y = np.random.binomial(1, 0.5, 2000)
```

```python
# Creating DataFrame
df = pd.DataFrame(X, columns=['Var1', 'Var2', 'Var3', 'Var4', 'Var5'])
y = pd.DataFrame(y, columns=['Target'])
```

```python
# Display Summary Statistics
print("Dataset Summary")
print(df.describe())
```

```python
# Histogram 
fig, axs = plt.subplots(2,3 , figsize =(10, 15))

sns.histplot(df["Var1"], ax = axs[0,0], bins = 20)
axs[0,0].set_title('Variable 1')

sns.histplot(df["Var2"], ax = axs[0,1], bins = 20)
axs[0,1].set_title('Variable 2')

sns.histplot(df["Var3"], ax = axs[0,2], bins = 20)
axs[0,2].set_title('Variable 3')

sns.histplot(df["Var4"], ax = axs[1,0], bins = 20)
axs[1,0].set_title('Variable 4')

sns.histplot(df["Var5"], ax = axs[1,1], bins = 20)
axs[1,1].set_title('Variable 5')

axs[1, 2].axis('off')

plt.tight_layout()
plt.show()
```

```python
# spliting Data
X_train, X_test, y_train, y_test = train_test_split(df, y, test_size = 0.3, random_state = 42)
```

```python
# Train Decision Tree
model = DecisionTreeClassifier(random_state = 42)
model.fit(X_train, y_train)
```

```python
# Make Prediction
y_pred = model.predict(X_test)
```

```python
# Visualizing Tree
plt.figure(figsize=(20,10))
plot_tree(model, feature_names=['Var1', 'Var2', 'Var3', 'Var4', 'Var5'], class_names=['Class 0', 'Class 1'], filled=True)
plt.show()
```

```python
# Evaluation
acc = accuracy_score(y_test, y_pred)

print("\n Accuracy: ")
print(acc)
```

```python
print("\n Confussion Matrix: ")
print(confusion_matrix(y_test, y_pred))
```

```python
print("\n Classification Report: ")
print(classification_report(y_test, y_pred))
```

## Reflection / Discussion
Why is the accuracy low?

The accuracy is usually close to 50% because:

The features were generated completely at random.
The target variable (y) was also generated randomly.
There is no relationship between the input variables and the target.
The model has no meaningful pattern to learn.
### How could accuracy be improved?
Use real-world data with meaningful relationships.
Create features that are correlated with the target variable.
Increase the quality of the data.
Perform feature engineering.
Tune hyperparameters such as:
max_depth
min_samples_split
min_samples_leaf
Use ensemble methods such as Random Forest, which is the next topic in the workshop.
