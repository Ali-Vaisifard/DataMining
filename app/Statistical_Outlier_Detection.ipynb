```python
import pandas as pd
import numpy as np
```

```python
# Activity 3 Solution

# Step 1: Load the dataset
df = pd.read_csv("../data/handling_anomalies_activity.csv")
```

```python
# Step 2: Inspect the dataset
print("\nFIRST 5 ROWS")
print(df.head())

print("\nDESCRIPTIVE STATISTICS")
print(df["Transaction_Value"].describe())
```

```python
# Part A - Z-Score Method


mean = df["Transaction_Value"].mean()
std = df["Transaction_Value"].std(ddof=1)

print("\nZ-SCORE CALCULATION")
print(f"Mean = {mean:.2f}")
print(f"Standard Deviation = {std:.2f}")
```

```python
# Calculate Z-Scores
df["Z_Score"] = (df["Transaction_Value"] - mean) / std
```

```python
# Threshold = |3|
z_outliers = df[np.abs(df["Z_Score"]) > 3]

print("\n Z-SCORE OUTLIERS")

if len(z_outliers) > 0:
    print(z_outliers)
else:
    print("No outliers detected.")
```

```python
# Part B - Modified Z-Score Method


median = df["Transaction_Value"].median()
mad = np.median(np.abs(df["Transaction_Value"] - median))

print("\n MODIFIED Z-SCORE  ")
print(f"Median = {median:.2f}")
print(f"MAD = {mad:.2f}")
```

```python
# Calculate Modified Z-Score
df["Modified_Z"] = 0.6745 * (
    (df["Transaction_Value"] - median) / mad
)
```

```python
# Threshold = |3.5|
modified_outliers = df[np.abs(df["Modified_Z"]) > 3.5]

print("\n MODIFIED Z-SCORE OUTLIERS")

if len(modified_outliers) > 0:
    print(modified_outliers)
else:
    print("No outliers detected.")
```

```python
# Compare Results
# 

print("\nCOMPARISON")

print("Outliers detected by Z-Score:")
print(z_outliers["Transaction_Value"].tolist())

print("\nOutliers detected by Modified Z-Score:")
print(modified_outliers["Transaction_Value"].tolist())
```

```python
# Summary

print("\n SUMMARY ")

print(f"Number of Z-Score outliers: {len(z_outliers)}")
print(f"Number of Modified Z-Score outliers: {len(modified_outliers)}")
```

## INTERPRETATION 
### The Z-Score method measures how many standard deviations
### each observation is from the mean.

The Modified Z-Score method uses the median and Median Absolute Deviation (MAD), making it less sensitive to extreme values.

For this dataset, the Modified Z-Score is generally more reliable because a few extreme transaction values (such as 10, 15, 200 and 220) influence the mean and standard deviation used by the standard Z-Score.

Since the majority of transaction values are concentrated around 90–110, the median provides a better measure of central tendency, allowing the Modified Z-Score to detect anomalies more robustly.
