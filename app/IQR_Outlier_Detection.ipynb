```python
import pandas as pd
```

```python
# Activity 2 Solution


# Step 1: Load the dataset
df = pd.read_csv("../data/handling_anomalies_activity.csv")
```

```python
# Step 2: Display the dataset
print("FIRST 5 ROWS")
print(df.head())

print("\nLAST 5 ROWS")
print(df.tail())

print("\nDESCRIPTIVE STATISTICS")
print(df.describe())
```

```python
# Step 3: Select the numerical column
column = "Transaction_Value"
```

```python
# Step 4: Calculate Q1, Q3 and IQR
Q1 = df[column].quantile(0.25)
Q3 = df[column].quantile(0.75)
IQR = Q3 - Q1

print("\nIQR CALCULATION")
print(f"Q1 = {Q1:.2f}")
print(f"Q3 = {Q3:.2f}")
print(f"IQR = {IQR:.2f}")
```

```python
# Step 5: Calculate the lower and upper bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"Lower Bound = {lower_bound:.2f}")
print(f"Upper Bound = {upper_bound:.2f}")
```

```python
# Step 6: Identify the outliers
outliers = df[
    (df[column] < lower_bound) |
    (df[column] > upper_bound)
]

print("\nOUTLIERS")
print(outliers)
```

```python
# Step 6: Identify the outliers
outliers = df[
    (df[column] < lower_bound) |
    (df[column] > upper_bound)
]

print("\nOUTLIERS")
print(outliers)
```

```python
# Step 8: Answer the activity question
print("\nACTIVITY QUESTION:")

if len(outliers) > 0:
    print("The outlier Record_IDs are:")
    print(outliers["Record_ID"].tolist())

    print("\nThe corresponding Transaction Values are:")
    print(outliers["Transaction_Value"].tolist())
else:
    print("No outliers were detected using the IQR method.")
```
