```python
import pandas as pd
import numpy as np
from sklearn.neighbors import LocalOutlierFactor
```

```python
# Activity 4 Solution

# Step 1: Load the dataset
df = pd.read_csv("../data/handling_anomalies_activity.csv")
```

```python
# Step 2: Display the first few rows
print("\n FIRST 5 ROWS ")
print(df.head())

print("\n DATASET INFORMATION  ")
print(df.info())
```

```python
# Step 3: Select numerical features only

numeric_df = df.select_dtypes(include=[np.number])

print("\n NUMERICAL FEATURES ")
print(numeric_df.head())
```

```python
# Step 4: Apply Local Outlier Factor (LOF)


lof = LocalOutlierFactor(
    n_neighbors=10,
    contamination=0.08
)

labels = lof.fit_predict(numeric_df)

# LOF labels:
#  1  = Normal observation
# -1  = Outlier

scores = lof.negative_outlier_factor_

# Add the results to the dataset
df["LOF_Label"] = labels
df["LOF_Score"] = scores
```

```python
# Step 5: Display detected outliers


outliers = df[df["LOF_Label"] == -1]

print("\n OUTLIERS DETECTED  ")
print(outliers)

print(f"\nNumber of outliers detected: {len(outliers)}")
```

```python
# Step 6: Remove outliers


cleaned_df = df[df["LOF_Label"] != -1]

print("\n CLEANED DATASET ")
print(cleaned_df.head())

print("\nOriginal dataset shape :", df.shape)
print("Cleaned dataset shape  :", cleaned_df.shape)
```

## INTERPRETATION 
### LOF compares the local density of each observation with the density of its nearest neighbours.

Observations with a LOF label of -1 are considered anomalies because they are located in regions of much lower density than the surrounding data.

Unlike the statistical methods, LOF does not assume that the data follows a normal distribution, making it suitable for real-world datasets.
