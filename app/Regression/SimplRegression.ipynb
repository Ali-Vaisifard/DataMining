```python
# import the librarirs
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
#load the dataset
df = pd.read_csv('../../data/Activity1LinearRegression.csv')
# display the dataframe
df.head() # print(df.head()) /  print(df.tail())
```

```python
#check for missing value
print(df.isnull().sum())
```

```python
# Display descriptive statistics
print(df.describe())
```

```python
# Scatterplot  of the data
plt.scatter(df['Year'], df['Per capita income (E)'])
plt.title("Scatterplot of x and y")
plt.xlabel("Year")
plt.ylabel("Per capita income (E)")
plt.show()
```

```python
# create the feature matrix 
X = df[['Year']] # double bracket creates a dataframe, single bracket creates a series
y = df['Per capita income (E)']
```

```python
# training the model
model = LinearRegression() # Linearregression.fit(X,y)
model.fit(X, y)
# make predictions
y_pred = model.predict(X)
```

```python
# model parameters
print('Intercept: ', model.intercept_)
print('Slope: ', model.coef_[0])
# y = Slope*X + Intercept
```

```python
# evaluate the model
rmse = np.sqrt(mean_squared_error(y, y_pred))
r2 = r2_score(y, y_pred)
print('RMSE: ', rmse)
print('R2 Score: ', r2)
```

```python
#predict for future#
future_year = [[2027]]
pred_income = model.predict(future_year)
print('Predicted income for 2027: ', pred_income[0])
# In 2027 Income is going to be 44267 + - 4001
```

```python
# plot
plt.Figure(figsize=(9,6))
plt.scatter(X, y, color='blue', label='Data points')
plt.plot(X, y_pred, color='red', linewidth=2, label='Linear Regression Line')
plt.xlabel("Year")
plt.ylabel("Per capita income (E)")
plt.title("Year vs Per Capita Income")
plt.legend()
plt.grid(True)
plt.show()
```
