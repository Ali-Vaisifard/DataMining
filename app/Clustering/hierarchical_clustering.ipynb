```python
import pandas as pd
import matplotlib.pyplot as plt 

from scipy.cluster.hierarchy import linkage, dendrogram, fcluster
from sklearn.preprocessing import StandardScaler
%matplotlib inline
```

```python
df = pd.read_csv("../hierarchical_clustering_dataset.csv")

print(df.head())
print(df.tail())
```

```python
print(df.describe())
```

```python
X = df[['Income_k', 'Spending_Score']]
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
linked = linkage(X_scaled, method='ward') # The other option for method are: single, average, completee, single, ceroid, median, weighted, and linkage, and 
```

```python
plt.figure(figsize=(10, 7))

dendrogram(
    linked,
    labels= df['Customer'].values
)

plt.title("Hierarchical Clustering Dendorogram")
plt.xlabel("Customers")
plt.ylabel("Euclidean distances")
plt.savefig('hierarchical_cluster.png', dpi= 300, bbox_inches='tight')
```
